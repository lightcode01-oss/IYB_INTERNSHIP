## Prerequisites

Install `requirements-dev.txt` (and `requirements-api.txt` for API work) in the active Python environment. Use the repository checkout and the saved Day 17 model. The setup below locates or clones the repository for Colab. Run cells in order.


```python
from pathlib import Path
import os, sys, subprocess, json
root = Path.cwd()
if not (root / "src").exists() and (root.parent / "src").exists():
    root = root.parent
if not (root / "src").exists():
    target = root / "IYB_INTERNSHIP"
    if not target.exists():
        subprocess.run(["git", "clone", "https://github.com/lightcode01-oss/IYB_INTERNSHIP.git", str(target)], check=True)
    root = target
os.chdir(root)
sys.path.insert(0, str(root))
print("Project:", root.name)
```

Run all application, API, schema, monitoring and explainability tests.

```python
r=subprocess.run([sys.executable,"-m","pytest","-q"],capture_output=True,text=True)
print(r.stdout)
assert r.returncode==0,r.stderr
```

Compile the source modules and application entry points.

```python
r=subprocess.run([sys.executable,"-m","compileall","-q","src","pages","app.py","api.py"],capture_output=True,text=True)
assert r.returncode==0,r.stderr
print("Application and source compilation: PASS")
```
