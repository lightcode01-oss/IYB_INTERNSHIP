## Theory

A dependency lock records transitive versions as well as direct libraries. An isolated environment prevents project setup from altering the global Python installation.

## Prerequisites

Install `requirements-dev.txt` (and `requirements-api.txt` for API work) in the active Python environment. Use the repository checkout and the saved Day 17 model. The setup below locates or clones the repository for Colab. Run cells in order.


```python
from pathlib import Path
import os, sys, subprocess, json
root = Path.cwd()
if not (root / "src").exists() and (root.parent / "src").exists():
    root = root.parent
if not (root / "src").exists():
    target = root / "IYB_INTERNSHIP"
    if not target.exists():
        subprocess.run(["git", "clone", "https://github.com/lightcode01-oss/IYB_INTERNSHIP.git", str(target)], check=True)
    root = target
os.chdir(root)
sys.path.insert(0, str(root))
print("Project:", root.name)
```

Check dependency consistency and the setup prerequisites.

```python
r=subprocess.run([sys.executable,"-m","pip","check"],capture_output=True,text=True)
print(r.stdout)
assert r.returncode==0,r.stderr
r=subprocess.run([sys.executable,"scripts/setup.py","--check"],capture_output=True,text=True)
print(r.stdout)
assert r.returncode==0,r.stderr
```

Inspect the captured versions and validate the reproducibility scripts.

```python
from importlib.metadata import version
print({p:version(p) for p in ["scikit-learn","pandas","numpy","nbclient"]})
lock=Path("requirements-lock.txt").read_text()
assert "scikit-learn==1.9.0" in lock
r=subprocess.run([sys.executable,"-m","py_compile","scripts/setup.py","scripts/execute_notebooks.py"],capture_output=True,text=True)
assert r.returncode==0,r.stderr
print("Lock and setup scripts: PASS")
```
